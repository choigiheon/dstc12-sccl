```python
from topclus.utils import TopClusUtils
utils = TopClusUtils()
utils.cluster(emb_path="results_dstc12-data/AppenBanking/latent_doc_emb.pt", n_clusters=20)
```

```python
import json
import os

# 클러스터링 결과 가져오기
cluster_results = utils.cluster(emb_path="results_dstc12-data/AppenBanking/latent_doc_emb.pt", n_clusters=20)

# 텍스트 파일 읽기
text_file_path = "dstc12-data/AppenBanking/texts.txt"
with open(text_file_path, "r", encoding="utf-8") as f:
    utterances = [line.strip() for line in f.readlines()]

# 결과 매핑 생성
cluster_label_map = {}
for i, (utterance, cluster_idx) in enumerate(zip(utterances, cluster_results)):
    cluster_label_map[utterance] = int(cluster_idx)

# JSON 파일로 저장
output_path = "cluster_label_map.json"
with open(output_path, "w", encoding="utf-8") as f:
    json.dump(cluster_label_map, f, ensure_ascii=False, indent=2)

print(f"클러스터 레이블 매핑이 {output_path}에 저장되었습니다.")
print(f"총 {len(cluster_label_map)}개의 발화가 {len(set(cluster_results))}개의 클러스터로 매핑되었습니다.")
```
